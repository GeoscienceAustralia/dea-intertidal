```python
import pandas as pd
import copy
import matplotlib.pyplot as plt
import numpy as np
```

Import the pandas.Series for the seagrass/macrophyte datasets from each of 
dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_case_TS_GoC.ipynb
dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_cases_GBR.ipynb
dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_cases_NSW.ipynb

Run those notebooks first then sequentially copy the datasets into this summary workflow

```python
# Run and output the GBR data from dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_cases_GBR.ipynb
%store -r data_GBR
# data_GBR = copy.deepcopy(seagrass)
data_GBR
```

```python
# Run and output the TS_GoC data from dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_case_TS_GoC.ipynb
%store -r data_TS_GoC
# data_TS_GoC = copy.deepcopy(seagrass)
data_TS_GoC
```

```python
# Merge the NESP datasets
data_NESP = pd.concat([data_GBR, data_TS_GoC], axis=1)
data_NESP['merged'] = data_NESP.apply(lambda row: row[data_NESP.columns[0]] + row[data_NESP.columns[1]], axis=1)

data_NESP
```

```python
# Run and import the NSW data from dea_intertidal/dea-intertidal/notebooks/experimental/Exposure_use_cases_NSW.ipynb
%store -r values_by_macrophyte
data_NSW = copy.deepcopy(values_by_macrophyte)
```

```python
# Merge the NSW and NESP datasets
merged3=pd.concat([data_NESP['merged'], data_NSW])
merged3
# merged.columns[2]
```

```python
# Bar plot the data

seagrass_counts = pd.Series(data = list(len(x) for x in merged3.array), index = merged3.index)

plt.figure(figsize=(25,7))
plt.boxplot(merged3, tick_labels=merged3.index)

for i, (name,count) in enumerate(seagrass_counts.items()):
    plt.text(i+1.25, 90, f'n={count}', ha='right', va = 'center', color='black',
             bbox=dict(facecolor='white', edgecolor='none', alpha=0.7))

plt.ylim(0,100)
```

```python
# Violin plot the data

data = list(merged3)
deciles = [np.percentile(values, np.arange(10,100,10)) for values in data]
# seagrass_counts = pd.Series(data = list(len(x) for x in merged3.array), index = merged3.index)

plt.figure(figsize=(25,7))
plt.violinplot(data)

for i, decile_values in enumerate(deciles, start=1):
    plt.scatter([i] * len(decile_values), decile_values, color='r', marker='o', s=15, label='Deciles' if i==1 else "")

for i, (name,count) in enumerate(seagrass_counts.items()):
    plt.text(i+1.25, 90, f'n={count}', ha='right', va = 'center', color='black',
             bbox=dict(facecolor='white', edgecolor='none', alpha=0.7))

plt.xticks(range(1,len(merged3) +1), merged3.index)
plt.ylim(0,100)

plt.xlabel("Seagrass sp.")
plt.ylabel("Exposure %")
plt.title("Data density distribution (Violin) plot of exposure\n values per seagrass species with deciles (red)")
```

```python
data = {
    'Cymodocea' : merged3.loc['Cymodocea\nrotundata'] + merged3.loc['Cymodocea\nserrulata'],
    'Enhalus' : merged3.loc['Enhalus\nacoroides'],
    'Halodule' : merged3.loc['Halodule\nuninervis'],
    'Halophila' : merged3.loc['Halophila\ncapricorni'] + merged3.loc['Halophila\ndecipiens'] + merged3.loc['Halophila\novalis'] + merged3.loc['Halophila\nspinulosa'] + merged3.loc['Halophila\ntricostata'],
    'Syringodium' : merged3.loc['Syringodium\nisoetifolium'],
    'Thalassia' : merged3.loc['Thalassia\nhemprichii'],
    'Thalassodendron' : merged3.loc['Thalassodendron\nciliatum'],
    'Zostera' : merged3.loc['Zostera\nmuelleri'] + merged3.loc['Zostera'],
    'Posidonia' : merged3.loc['Posidonia'],
    'Posidonia_Zostera' : merged3.loc['Posidonia/Zostera'],
    'Mangrove' : merged3.loc['Mangrove'],
    'Saltmarsh' : merged3.loc['Saltmarsh'],
    }
merged4 = pd.Series(data=data)
merged4
```

```python
# Bar plot the data
seagrass_counts = pd.Series(data = list(len(x) for x in merged4.array), index = merged4.index)

plt.figure(figsize=(25,7))
plt.boxplot(merged4, tick_labels=merged4.index)

for i, (name,count) in enumerate(seagrass_counts.items()):
    plt.text(i+1.1, 98, f'n={count}', ha='right', va = 'center', color='black',
             bbox=dict(facecolor='white', edgecolor='none', alpha=0))

plt.ylim(0,100)
```

```python
# Violin plot the data

data = list(merged4)
deciles = [np.percentile(values, np.arange(10,100,10)) for values in data]
# seagrass_counts = pd.Series(data = list(len(x) for x in merged3.array), index = merged3.index)

plt.figure(figsize=(25,7))
plt.violinplot(data)

for i, decile_values in enumerate(deciles, start=1):
    plt.scatter([i] * len(decile_values), decile_values, color='r', marker='o', s=15, label='Deciles' if i==1 else "")

for i, (name,count) in enumerate(seagrass_counts.items()):
    plt.text(i+1.1, 98, f'n={count}', ha='right', va = 'center', color='black',
             bbox=dict(facecolor='white', edgecolor='none', alpha=0.0))

plt.xticks(range(1,len(merged4) +1), merged4.index)
plt.ylim(0,100)

plt.xlabel("Seagrass")
plt.ylabel("Exposure %")
plt.title("Data density distribution (Violin) plot of exposure\n values per seagrass species with deciles (red)")
```
